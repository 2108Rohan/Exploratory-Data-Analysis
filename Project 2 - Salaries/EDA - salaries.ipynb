## Replace 'Not Provided' in EmployeeName column to NaN

```python
data['EmployeeName']= data['EmployeeName'].replace('Not provided',np.nan)
```

## Drop the rows having 1 missing values

```python
data=data.drop(data[data.isnull().sum(axis=1)==1].index,axis=0)
# we are using index to give list of indexes to drop function as 1st parameter
```

## Find job title of Albert Pardini

## Find Average BasePay of All Employee Per Year 

```python
data.groupby('Year')['BasePay'].mean()
```

## Find Average BasePay of All Employee Per JobTitle 

```python
data.groupby('JobTitle')['BasePay'].mean()
```

## Find Average BasePay of Employee Having Job Title ACCOUNTANT  

```python
data[data['JobTitle'].str.contains('Accountant')]['BasePay'].mean()
```

## Display Name of The Person Having The Highest BasePay

```python
data[data['BasePay'] == data['BasePay'].max()]['EmployeeName']
# data['BasePay'] == data['BasePay']....this produces a boolean array which we can pass in data
```
